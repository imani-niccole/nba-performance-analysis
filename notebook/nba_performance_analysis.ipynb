# Moving into analysis
### Striving to answer: Are NBA teams associated with elite individual performance or balanced contributions across players?

Let's start with basic team metric summaries:


```python
"""
 groupby = group the 578 player records by NBA team name and abbreviation
 agg = for each player, calculate the following stats
"""

team_player_summary = (
    players_df
    .groupby(["Team", "Team_Name"])
    .agg(
        player_count = ("Player_Name", "count"),
        avg_player_ppg = ("Points", "mean"),
        avg_player_assists = ("Assists", "mean"),
        avg_player_rebounds = ("Rebounds", "mean")
    )
    .reset_index()
)
team_player_summary.head()
```

```python
team_player_summary.shape
```

### Finding average player PPG

```python
# now we want to sort teams based on average player ppg (points per game)

team_player_summary.sort_values(
    "avg_player_ppg",
    ascending=False
).head(10)
```

### Finding the top scorers for each team

```python
# sort players by points, the first row is the top scorer for respective team
top_scorers = (
    players_df
    .sort_values("Points", ascending=False)
    .groupby("Team_Name")
    .first()
    .reset_index()
)

# renaming the columns
top_scorers = top_scorers[
    ["Team_Name", "Player_Name", "Points"]
].rename(columns={
    "Player_Name": "Top_Scorer",
    "Points": "Top_Scorer_PPG"
})

# nowwww, sort from highest to lowest player ppg
top_scorers = top_scorers.sort_values(
    "Top_Scorer_PPG",
    ascending=False
).reset_index(drop=True)

top_scorers.head(10)
```

# Core Metric: Scoring Concentration
### Calculate Top-3 Scoring
Identify the three highest-scoring players by PPG:

```python
top_3_scoring = (
    players_df
    .sort_values(
        ["Team_Name", "Points"],
        ascending=[True,False]

    )
    .groupby("Team_Name")
    .head(3)
)
top_3_scoring.head(10)
```

```python
# add top-3 ppg values

top_3_summary = (
    top_3_scoring
    .groupby("Team_Name")
    .agg(
        Top_3_PPG = ("Points", "sum")
    )
    .reset_index()
)

top_3_summary = (
    top_3_summary
    .sort_values("Top_3_PPG", ascending=False)
    .reset_index(drop=True)
)

top_3_summary.head(10)
```

### Calculating Scoring Concentration
### Formula: ((Top 3 combined PPG) / (Top 10 combined PPG)) x 100
**Meaning:** A higher percentage means scoring is more concentrated among the top 3 scorers. A lower percentage means scoring is more concentrated acorss the top 10 scorers

### Get the top 10 players

```python
top_10_scoring = (
    players_df
    .sort_values(
        ["Team_Name", "Points"],
        ascending=[True, False]
    )
    .groupby("Team_Name")
    .head(10)
)
```

```python
top_10_summary = (
    top_10_scoring
    .groupby("Team_Name")
    .agg(
        Top_10_PPG = ("Points", "sum")
    )
    .reset_index()
)
```

### Combine scoring summaries

```python
# combining 3 dataframes into 1

scoring_summary = (
    top_scorers
    .merge(top_3_summary, on="Team_Name")
    .merge(top_10_summary, on="Team_Name")
)

scoring_summary.head()
```

# -------------------------------------------------------------------------
# Calculating Scoring Concentration!
**Formula:** ((Top 3 combined PPG) / (Top 10 combined PPG)) x 100

```python
scoring_summary["Scoring_Concentration"] = (
    scoring_summary["Top_3_PPG"] /
    scoring_summary["Top_10_PPG"]
) * 100

scoring_summary["Scoring_Concentration"] = (
    scoring_summary["Scoring_Concentration"].round(2)
)
```

# NOTE!
### Don't confuse scoring concentration as the percentage of the team's actual points, but the sum of individual PPG avgs.
Higher % --> scoring is more concentrated among top 3 players  
Lower % --> scoring is more distributed among top 10 players

```python
scoring_summary.sort_values(
    "Scoring_Concentration",
    ascending=False
).head(10)
```

### Most Concentrated

```python
# looking at the two extremes

scoring_summary.nlargest(
    5,
    "Scoring_Concentration"
)[
    ["Team_Name", "Top_Scorer", "Top_Scorer_PPG",
     "Top_3_PPG", "Top_10_PPG", "Scoring_Concentration"]
]
```

### Most Balanced

```python
scoring_summary.nsmallest(
    5,
    "Scoring_Concentration"
)[
   ["Team_Name", "Top_Scorer", "Top_Scorer_PPG",
    "Top_3_PPG", "Top_10_PPG", "Scoring_Concentration"]
]
```

### Merge with team performance

```python
nba_analysis = teams_df.merge(
    scoring_summary,
    on="Team_Name",
    how="left"
)

nba_analysis.head()
```

```python
# calculate the correlation coefficient between 3 variables
nba_analysis[
    ["WIN%", "Scoring_Concentration", "Top_Scorer_PPG"]
].corr()
```

### What the Data Says ~ Correlation Coefficient 
### Focusing on the relationship between Win Percentage and Scoring Concentration = 0.716467

**Positive:** higher-performing teams tend to have more concentrated top-3 scoring.
**Negative:** higher-performing teams tend to have more distributed scoring.
**Near Zero:** scoring concentratoin doesn't have a strong relationship with winning

```python
win_concentration_corr = nba_analysis[
    "WIN%"
].corr(
    nba_analysis["Scoring_Concentration"]
)

win_top_corr = nba_analysis[
    "WIN%"
].corr(
    nba_analysis["Top_Scorer_PPG"]
)

print(
    f"Win % vs. Scoring Concentration: "
    f"{win_concentration_corr:.3f}"
)


print(
    f"Win % vs. Top Scorer PPG: "
    f"{win_top_corr:.3f}"
)
```

### Scoring Findings

Scoring concentration has a **strong positive correlation** with win percentage (r = 0.716), while top-scorer PPG has a moderate positive correlation with win percentage (r = 0.556).

Through this dataset, team success was more strongly associated with scoring among the top three players than the team's leading scorer alone.

These relationships are correlational, not causational.

### Assist Concentration
**Formula:** Assist Concentration = ((Top 3 players' APG) / (Top 10 players' APG)) x 100  
Higher = assists are concentrated among a few players  
Lower = assists are balanced among top 10.

```python
top_3_assists = (
    players_df
    .sort_values(
        ["Team_Name", "Assists"],
        ascending=[True, False]
    )
    .groupby("Team_Name")
    .head(3)
)

top_10_assists = (
    players_df
    .sort_values(
        ["Team_Name", "Assists"],
        ascending=[True, False]
    )
    .groupby("Team_Name")
    .head(10)
)
```

```python
assist_summary = (
    top_3_assists
    .groupby("Team_Name")
    .agg(Top_3_APG = ("Assists", "sum"))
    .reset_index()
)

top_10_assist_summary = (
    top_10_assists
    .groupby("Team_Name")
    .agg(Top_10_APG = ("Assists", "sum"))
    .reset_index()
)
```

```python
# merge

assist_summary = assist_summary.merge(
    top_10_assist_summary,
    on = "Team_Name")
```

```python
assist_summary["Assist_Concentration"] = (
    assist_summary["Top_3_APG"] /
    assist_summary["Top_10_APG"]
) * 100

assist_summary["Assist_Concentration"] = (
    assist_summary["Assist_Concentration"].round(2)
)

assist_summary.sort_values(
    "Assist_Concentration",
    ascending=False
).head(10)
```

```python
nba_analysis = nba_analysis.merge(
    assist_summary,
    on="Team_Name",
    how="left"
)
```

```python
nba_analysis[
    ["WIN%", "Team_Assists", "Assist_Concentration"]
].corr()
```

### What are the numbers saying

Moderate positive relationship between assist concentration and win percentage (r = 0.481)

### Testing Rebounds

```python
top_3_rebounds = (
    players_df
    .sort_values(
        ["Team_Name", "Rebounds"],
        ascending = [True, False]
    )
    .groupby("Team_Name")
    .head(3)
)

top_10_rebounds = (
    players_df
    .sort_values(
        ["Team_Name", "Rebounds"],
        ascending = [True, False]
    )
    .groupby("Team_Name")
    .head(10)
)
        
```

```python
rebound_summary = (
    top_3_rebounds
    .groupby("Team_Name")
    .agg(Top_3_RPG = ("Rebounds", "sum"))
    .reset_index()
)

top_10_rebound_summary = (
    top_10_rebounds
    .groupby("Team_Name")
    .agg(Top_10_RPG = ("Rebounds", "sum"))
    .reset_index()
)

rebound_summary = rebound_summary.merge(
    top_10_rebound_summary,
    on = "Team_Name"
)
```

```python
rebound_summary["Rebound_Concentration"] = (
    rebound_summary["Top_3_RPG"] / 
    rebound_summary["Top_10_RPG"]
) * 100

rebound_summary["Rebound_Concentration"] = (
    rebound_summary["Rebound_Concentration"].round(2)
)
```

```python
# merge into dataset

nba_analysis = nba_analysis.merge(
    rebound_summary,
    on = "Team_Name",
    how = "left"
)
```

```python
nba_analysis[
    ["WIN%", "Team_Rebounds", "Rebound_Concentration"]
].corr()
```

# Key Question: Are team rankings in the NBA driven by elite individual players or balanced team wide statistical contributors?
### Key Findings - Stars *do* run the league!
~ Teams with scoring concentrated at the top three players tended to have a higher win percentage (r = 0.716).  
~ A top-3 scoring concentration is more closely associated with winning than one elite player. The Top Scorer's PPG was r = 0.556.  

### Secondary Question: Are winning teams generally better at assisting and rebounding?
~ Assist Concentration (r = 0.481) has a stronger relationship with win percentage than Rebound Concentration (r = 0.354) and win percentage.  
~ In relation to win percentage, the results lean more towards assists  among three players (r = 0.481) than the overall team (r = 0.071).  
~ In relation to win percentage, the results lean more towards rebounds in overall team production (r = 0.511) than concentration among three players (r = 0.354).   

*NOTE:* These findings describe associations within the dataset and don't establish that player production causes NBA teams to win.
